```python
from dataclasses import dataclass
from typing import List, Tuple

@dataclass
class Node:
  name: str
  options: List[Tuple[str, 'Node']]

def gini(x: pd.Series | pd.DataFrame):
  p = x.value_counts(normalize=True)
  return 1 - np.sum(p**2)

def find_root_node(X: pd.DataFrame, y: pd.DataFrame | pd.Series):
  N = len(X)
  best_split = (float('inf'), None, None)

  for col in X:
    groups = X[col].unique()
    for group in groups:
      left = X[col] == group
      right = X[col] != group
      w = len(left) / N * gini(y[left]) + len(right) / N * gini(y[right])
      best_split = min(best_split, (w, col, group), key=lambda k: k[0])

  return best_split[1:]

tree = find_root_node(df.drop(columns=['Student', 'Liked']), df['Liked'])
tree
```
